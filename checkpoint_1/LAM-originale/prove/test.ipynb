```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 100)
y = x**3 - 3*x

plt.plot(x, y, color='blue', label='f(x) = x^3 - 3x')
plt.axhline(0, color='black', lw=0.8)
plt.axvline(0, color='black', lw=0.8)
plt.title('Funzione cubica')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
def fattoriale(numero):
    risultato = 1
    while numero > 1:
        risultato = risultato * numero
        numero = numero - 1
    return risultato

print(fattoriale(5))
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2*np.pi, 200)
y1 = np.sin(x)
y2 = 2*np.sin(x)
y3 = np.sin(2*x)

plt.plot(x, y1, color='blue', label='f(x) = sin(x)')
plt.plot(x, y2, color='red', label='f(x) = 2sin(x)')
plt.plot(x, y3, color='green', label='f(x) = sin(2x)')
plt.axhline(0, color='black', lw=0.8)
plt.axvline(0, color='black', lw=0.8)
plt.title('Funzioni trigonometriche')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
```
